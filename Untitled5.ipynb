```python
import pandas as pd

df = pd.read_excel("Curriculum_Industry_Skill_Alignment_1000.xlsx")

print(df.head())
```

```python
print("Dataset Shape:", df.shape)

print("\nColumns:")
print(df.columns.tolist())

print("\nDataset Information:")
df.info()
```

```python
print(df.isnull().sum())
```

```python
print(df["Is_Employable"].value_counts())
print(df["Is_Employable"].value_counts(normalize=True))
```

```python
print(df["Is_Employable"].value_counts(normalize=True))
```

```python
features = [
    "Overall_CGPA",
    "DSA_Grade",
    "DBMS_Grade",
    "OOP_Grade",
    "OS_Grade",
    "LeetCode_Solved_Count",
    "GitHub_Repo_Count",
    "Has_Internship_Exp",
    "Internship_Duration_Months",
    "Hackathon_Participation_Count",
    "Has_Cloud_Certification",
    "Primary_Tech_Stack",
    "Quantitative_Aptitude_Score",
    "Logical_Reasoning_Score",
    "Coding_Automata_Score",
    "Communication_Skill_Rating",
    "Knows_Git_Version_Control",
    "Knows_Python",
    "Knows_SQL",
    "Knows_JavaScript"
]

X = df[features]

y = df["Is_Employable"]

print("Features shape:", X.shape)
print("Target shape:", y.shape)
```

```python
numerical_features = [
    "Overall_CGPA",
    "DSA_Grade",
    "DBMS_Grade",
    "OOP_Grade",
    "OS_Grade",
    "LeetCode_Solved_Count",
    "GitHub_Repo_Count",
    "Internship_Duration_Months",
    "Hackathon_Participation_Count",
    "Quantitative_Aptitude_Score",
    "Logical_Reasoning_Score",
    "Coding_Automata_Score",
    "Communication_Skill_Rating"
]
```

```python
categorical_features = [
    "Primary_Tech_Stack"
]
```

```python
numerical_pipeline = Pipeline(
    steps=[
        ("missing_values", SimpleImputer(strategy="median")),
        ("scaling", StandardScaler())
    ]
)
```

```python
categorical_pipeline = Pipeline(
    steps=[
        ("missing_values", SimpleImputer(strategy="most_frequent")),
        ("encoding", OneHotEncoder(handle_unknown="ignore"))
    ]
)
```

```python
preprocessor = ColumnTransformer(
    transformers=[
        (
            "numerical",
            numerical_pipeline,
            numerical_features
        ),
        (
            "categorical",
            categorical_pipeline,
            categorical_features
        )
    ]
)
```

```python
model_pipeline = Pipeline(
    steps=[
        ("preprocessing", preprocessor),
        ("model", LogisticRegression(max_iter=1000))
    ]
)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)
```

```python
model_pipeline.fit(X_train, y_train)
```

```python
predictions = model_pipeline.predict(X_test)

print("First 20 Predictions:")
print(predictions[:20])
```

```python
accuracy = accuracy_score(y_test, predictions)

print("Accuracy:", accuracy)
print("Accuracy Percentage:", accuracy * 100)
```

```python
print("Classification Report:")
print(classification_report(y_test, predictions))
```

```python
cm = confusion_matrix(y_test, predictions)

print("Confusion Matrix:")
print(cm)
```
